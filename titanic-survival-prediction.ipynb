```python
dft = pd.concat([df1, df2])
```

```python
dft.head()
```

```python
dft.shape
```

```python
dft.isnull().sum()
```

```python
dft.describe()
```

```python
dft.corr(numeric_only=True)
```

```python
dft.columns
```

```python
dft=dft.drop(['Ticket','PassengerId','Cabin'], axis=1)
```

```python
dft.head()
```

```python
dft.shape
```

```python
dft.isnull().sum()
```

```python
dft['Fare'].value_counts()
```

```python
dft['Fare']=dft['Fare'].fillna(dft['Fare'].mean())
```

```python
dft.isnull().sum()
```

```python
dft['Embarked'].value_counts()
```

```python
dft['Embarked']=dft['Embarked'].fillna(dft['Embarked'].mode()[0])
```

```python
dft.isnull().sum()
```

```python
dft.groupby('Pclass')['Age'].median()
```

```python
dft['Age'] = dft['Age'].fillna(dft.groupby('Pclass')['Age'].transform('median'))
```

```python
dft.isnull().sum()
```

```python
dft.head()
```

```python
dft = dft.drop(['Name', 'Embarked'], axis=1)
```

```python
dft.head()
```

```python
dft = pd.get_dummies(dft, drop_first=True)
```

```python
dft.head()
```

```python
df1.shape
```

```python
train=dft[ :891]
test=dft[891: ]
```

```python
x_train=train.drop('Survived',axis=1)
y_train=train['Survived']
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
```

```python
r=RandomForestClassifier()
g=GradientBoostingClassifier()
```

```python
r.fit(x_train,y_train)
```

```python
test=test.drop('Survived',axis=1)
```

```python
tahmin=r.predict(test)
```

```python
tahmin
```

```python
result=pd.DataFrame()
```

```python
result['Survived']=tahmin
result['PassengerId']=df2['PassengerId']
```

```python
result['Survived']=result['Survived'].astype('int32')
```

```python
result
```

```python
result.to_csv('result1.csv',index=False)
```
